Step 1 : Import Library

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Step 2 : Load Dataset

```python
df = pd.read_csv(r"C:\Users\Jagrat Darji\Desktop\Data Analytics\iphone Sales Analysis\Dataset\apple_products.csv")
print(df.info())
print(df.head())
```

Step 3 : Check for missing values

```python
print("\nMissing values:\n", df.isnull().sum())
```

Step 4 : Summary statistics

```python
print("\nSummary statistics:\n", df.describe())
```

Step 5 : Price Distribution

```python
plt.figure(figsize=(8,4))
sns.histplot(df['Sale Price'], bins=20, kde=True, color='green')
plt.title('price Distribution of Apple Products')
plt.xlabel('price')
plt.ylabel('frequency')
plt.show()
```

Step 6 : Bar Plot: Models vs Ratings

```python
plt.figure(figsize=(10,5))
sns.barplot(x='Sale Price', y='Star Rating', data=df.sort_values(by='Star Rating', ascending=False))
plt.xticks(rotation=90)
plt.title('Average Rating by iPhone Model')
plt.show()
```

Step 7 : Relationship between Price and Rating

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x='Sale Price', y='Star Rating', data=df, hue='Product Name')
plt.title('Star Rating vs Star Rating')
plt.show()
```

Step 8 : Count of Reviews per Model

```python
plt.figure(figsize=(10,4))
sns.barplot(x='Product Name', y='Number Of Reviews', data=df.sort_values(by='Number Of Reviews', ascending=False))
plt.xticks(rotation=90)
plt.title('Review Count by iPhone Model')
plt.show()
```
