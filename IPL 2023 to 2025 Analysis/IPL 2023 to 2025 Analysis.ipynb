Step 1: Import Library

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Step 2 : Load your Datasets

```python
batsman_df = pd.read_csv(r"C:\Users\Jagrat Darji\Desktop\Data Analytics\IPL 2023 to 2025 Analysis\Dataset\IPL2023_Batsman.csv", encoding='ISO-8859-1')
bowler_df = pd.read_csv(r"C:\Users\Jagrat Darji\Desktop\Data Analytics\IPL 2023 to 2025 Analysis\Dataset\IPL2023_Bowler.csv", encoding='ISO-8859-1')
```

Step 3: Top 10 Batsmen by Runs

```python
top_batsmen = batsman_df.groupby("Batsman")["Run"].sum().sort_values(ascending=False).head(10)
```

Step 4 : Top 10 Bowlers By Wickets 

```python
top_bowlers = bowler_df.groupby("Bowler")["wicket"].sum().sort_values(ascending=False).head(10)
```

Step 5 : Plotting

```python
plt.figure(figsize=(16, 6))
```

Step 6 : Batsmen Barplot

```python
plt.subplot(1, 2, 1)
sns.barplot(x=top_batsmen.values, y=top_batsmen.index, palette="YlGnBu")
plt.title("Top 10 Batsmen by Runs")
plt.xlabel("Total Runs")
plt.ylabel("Batsman")
```

Step 7 : Bowlers Barplot

```python
plt.subplot(1, 2, 2)
sns.barplot(x=top_bowlers.values, y=top_bowlers.index, palette="OrRd")
plt.title("Top 10 Bowlers by Wickets")
plt.xlabel("Total Wickets")
plt.ylabel("Bowler")
```

```python
plt.tight_layout()
plt.show()
```

Step 8 : Top 10 Batsmen by Strike Rate and Top 10 Bowler by Economy Rate

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load datasets
batsman_df = pd.read_csv(r"C:\Users\Jagrat Darji\Desktop\Data Analytics\IPL 2023 to 2025 Analysis\Dataset\IPL2023_Batsman.csv", encoding='ISO-8859-1')
bowler_df = pd.read_csv(r"C:\Users\Jagrat Darji\Desktop\Data Analytics\IPL 2023 to 2025 Analysis\Dataset\IPL2023_Bowler.csv", encoding='ISO-8859-1')

# ----------------------------
# 🏏 Top 10 Batsmen by Strike Rate
# ----------------------------
batsman_stats = batsman_df.groupby("Batsman").agg({"Run": "sum", "Ball": "sum"}).reset_index()
batsman_stats["Strike Rate"] = (batsman_stats["Run"] / batsman_stats["Ball"]) * 100
top_sr_batsmen = batsman_stats[batsman_stats["Ball"] > 30].sort_values(by="Strike Rate", ascending=False).head(10)

# ----------------------------
# 🎯 Top 10 Bowlers by Economy Rate
# ----------------------------
bowler_stats = bowler_df.groupby("Bowler").agg({"run": "sum", "over": "sum"}).reset_index()
bowler_stats["Economy"] = bowler_stats["run"] / bowler_stats["over"]
top_economy_bowlers = bowler_stats[bowler_stats["over"] > 8].sort_values(by="Economy").head(10)

# ----------------------------
# 📊 Plotting Both Charts
# ----------------------------
plt.figure(figsize=(16, 6))

# Strike Rate Plot
plt.subplot(1, 2, 1)
sns.barplot(x="Strike Rate", y="Batsman", data=top_sr_batsmen, palette="YlOrBr")
plt.title("Top 10 Batsmen by Strike Rate")
plt.xlabel("Strike Rate")
plt.ylabel("Batsman")

# Economy Plot
plt.subplot(1, 2, 2)
sns.barplot(x="Economy", y="Bowler", data=top_economy_bowlers, palette="BuPu")
plt.title("Top 10 Bowlers by Economy Rate")
plt.xlabel("Economy Rate")
plt.ylabel("Bowler")

plt.tight_layout()
plt.show()
```
